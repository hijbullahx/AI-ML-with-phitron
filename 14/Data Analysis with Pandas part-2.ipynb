# 14.1 Filtering Based on String

```python
import pandas as pd

emp_data = {
    'Name': [
        'Niloy', 'Ishrat', 'Pavel', 'Tuhin', 'Sadia', 'Morshed', 'Labiba', 'Harun'
    ],
    'City': [
        'Dhaka', 'new Chittagong', 'Khulna', 'Rajshahi', 'Sylhet', 'Barisal', 'Comilla', 'Mymensingh'
    ],
    'Department': [
        'HR', 'IT', 'Finance', 'Sales', 'Marketing', 'Support', 'Admin', 'IT'
    ],
    'Salary': [
        54000, 62000, 50000, 48000, 56000, 51000, 53000, 60000
    ]
}

df = pd.DataFrame(emp_data)
print(df)
```

```python
#filtering city name with n
df.loc[df['City'].str.contains("N", case = False)]
```

```python
#filtering starts with new
df.loc[df['City'].str.contains(r"^new")]
```

```python
#Ends with a
df.loc[df['City'].str.contains(r"a$")]
```

```python
# name starts with a vowel
df.loc[df['Name'].str.contains(r"^[AEIOU]", case = False)]
```

```python
#City contains c | d
df.loc[df['City'].str.contains(r"c|d", case = False)]
```

# 14.4 Handling Duplicates Values

```python
#check if duplicated
df.duplicated().sum()
```

```python
#Deketing duplicate
df.drop_duplicates()
```

```python
df
```

```python
#permanent delete

df.drop_duplicates(inplace=True)
```

```python
df
```

```python
#duplicate name based delete
df.drop_duplicates(subset=['Name'])
```

```python
#cities
df.drop_duplicates(subset=['City'], keep = 'last')
```

# 14.7 Apply Function in DF


```python
df
```

```python
#Min max scalling, scalled = (x-min)/(max-min)
mn = df['Total Marks'].min()
mx = df['Total Marks'].max()

df['Scaled Marks'] = df['Total Marks'].apply(lambda x : (x-mn)/(mx-mn))
```

```python
df
```

```python
#Custom build fuction
def grading_system(marks):
  if marks >=260:
    return 'A+'
  elif marks >= 250:
    return 'A'
  else:

    return 'A-' 

df['Grade']  = df['Total Marks'].apply(grading_system)
df
```

```python
def marking_system(df):
  a = df['Data Structure Marks'] * 2
  b = df['Python Marks'] * 3
  c = df['Algorithm Marks'] * 4

  return a+b+c


df['Exceptional Marks'] = df.apply(marking_system,axis=1)
df
```
