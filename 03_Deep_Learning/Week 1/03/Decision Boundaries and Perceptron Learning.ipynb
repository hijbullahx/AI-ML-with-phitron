1. What is a decision boundary, and how does it relate to classification tasks?

->
Definition: A decision boundary is the line (in 2D), plane (in 3D), or hyperplane (in higher dimensions) that separates different classes in the feature space.

Relation to Classification: It determines which side of the boundary a data point falls on, thereby deciding its predicted class.

2. How does the bias term in a neuron affect the position of the decision boundary?

->Effect: The bias shifts the decision boundary away from the origin.

Without bias: The boundary always passes through the origin.

With bias: The boundary can move up/down or left/right, allowing more flexible classification.



3. Explain with an example how changing weights affects the orientation of the decision boundary.

->Example:

Equation: 𝑤1 𝑥 1 + 𝑤2 𝑥2+ 𝑏 = 0


If 𝑤 = [ 1 , 1 ], boundary is 𝑥1 +𝑥2 + 𝑏 = 0
 → diagonal line.
If 𝑤 = [ 2, −1], boundary is 2 𝑥 1− 𝑥 2 + 𝑏 = 0
 → steeper line, rotated orientation.

3. Why is it important to visualize the decision boundary when training a model?'

->Reasons:

Helps understand how the model separates classes.

Reveals underfitting/overfitting.

Shows whether classes are linearly separable.

Aids in debugging and improving model design.
