# Practice:

1. What is bias in a neural network?

-> Bias is an extra parameter added to the weighted sum of inputs in a perceptron. It allows the model to shift the activation function left or right, giving flexibility in learning.

2. Why do we add bias to a perceptron?

-> Without bias, the output always passes through the origin (0,0). Bias ensures the perceptron can fit data better by adjusting the threshold at which the neuron activates.

3. What are weights in a perceptron?  

-> Weights are parameters that determine the importance of each input. Each input is multiplied by its weight before summation.

4. How do weights affect the output of a perceptron?

-> Larger weights amplify the input’s influence, while smaller weights reduce it. Negative weights can even reverse the effect of an input.

6. What is Step Function? How does it work?

->A simple activation function that outputs 1 if the input is above a threshold, otherwise 0. Example: If threshold = 0, then input ≥ 0 → output = 1; input < 0 → output = 0. It’s like an on/off switch.

7. Why do we need Activation Function?

-> Without activation functions, perceptrons would only produce linear outputs. Activation functions introduce non-linearity, allowing neural networks to learn complex patterns.

Three common activation functions:

Sigmoid: Outputs values between 0 and 1.

ReLU (Rectified Linear Unit): Outputs 0 for negative inputs, and the input itself for positive values.

Tanh: Outputs values between -1 and 1.
