# 10.3 Hands on of Linear Regression and Best Fitted Line


```python
# Step 1: mport Libraries
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.linear_model import LinearRegression
```

```python
# Step 2: Create a Simple Dataset
hours = np.array([2,4,6,8,10]).reshape(-1,1)
marks = np.array([55,60,70,80,88])
hours, marks
```

```python
#Step 3: Train a Linear Regression Model
#we use scikit learn to train the model
#the model will automatically learn: slop(m) and intersect(c)
model = LinearRegression()
model.fit(hours, marks)
```

```python
#Step 4: View the learned Parameters
#The regression line equation is: (y = mc + c), where, m = slope, c = intercept
print(f"Slope(m): {model.coef_[0]}")
print(f"Intercept(c): {model.intercept_}")

#So the equtaion is : marks = 4.3 X hours + 44.8
```

```python
# Step 5: Visualize the best fit line
plt. scatter(hours, marks, color = "blue", label = "Actual Data")

predicted_marks = model.predict(hours)
plt.plot(hours, predicted_marks, color = "red", label = "Best Fit Line")

plt.xlabel("Hourse")
plt.ylabel("Marks")
plt.title("Study Hours vs Marks for Linear Regression")
plt.legend()
plt.grid(True)
plt.show()
```

```python
#Predict Mark for a new Student
new_hours = np.array([[7]])
prediction = model.predict(new_hours)
print(f"Predicted Mark for {new_hours} hours: {prediction[0]}")
```

# 10.5 Hands on_Cost Function and Gradient Descent

```python
# Step 1: mport Libraries
from sklearn.linear_model import LinearRegression, SGDRegressor
from sklearn.metrics import mean_squared_error
# Step 2: Create a Simple Dataset
```

```python
# Step 3: fit a linear regression model(closed-form solver)
lin_reg = LinearRegression()
lin_reg.fit(hours, marks)

print(f"Slope(m): {lin_reg.coef_[0]}")
print(f"Intercept(c): {lin_reg.intercept_}")
```

```python
# Srep 4: Compute the Cost (MSE) Using a Library
pred_marks = lin_reg.predict(hours)
mse_lin = mean_squared_error(marks, pred_marks)
print(f"Mean Squared Error (MSE) using Linear Regression: {mse_lin}")
```

```python
# Step 5: Visualize the best fit line
plt. scatter(hours, marks, color = "blue", label = "Actual Data")

predicted_marks = model.predict(hours)
plt.plot(hours, pred_marks, color = "red", label = "Best Fit Line")

plt.xlabel("Hourse")
plt.ylabel("Marks")
plt.title("Study Hours vs Marks for Linear Regression")
plt.legend()
plt.grid(True)
plt.show()
```

```python
#Step 6: Use a Gradient_Descent_Based Model - SDGRegressor
sgd_reg = SGDRegressor(
    max_iter= 1000,
    learning_rate = "invscaling",
    eta0= 0.01,
    random_state=42
)

sgd_reg.fit(hours, marks)
print(f"Slope(m): {sgd_reg.coef_[0]}")
print(f"Intercept(c): {sgd_reg.intercept_[0]}")

pred_sgd = sgd_reg.predict(hours)
mse_sgd = mean_squared_error(marks, pred_sgd)
print(f"Mean Squared Error (MSE) using SGDRegressor: {mse_sgd}")
```

```python
from matplotlib.lines import lineStyles
# Step 7: Compare Both Line Visually
plt. scatter(hours, marks, color = "blue", label = "Actual Data")

predicted_marks = model.predict(hours)
plt.plot(hours, pred_marks, color = "red", label = "Linear Regression", linestyle = "--")
plt.plot(hours, pred_sgd, color = "orange", label = "SGD Regression", linestyle = "-.")

plt.xlabel("Hourse")
plt.ylabel("Marks")
plt.title("Linear Regression vsSGD Regression")
plt.legend()
plt.grid(True)
plt.show()
```
