# 10.3 Hands on of Linear Regression and Best Fitted Line


```python
# Step 1: mport Libraries
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.linear_model import LinearRegression
```

```python
# Step 2: Create a Simple Dataset
hours = np.array([2,4,6,8,10]).reshape(-1,1)
marks = np.array([55,60,70,80,88])
hours, marks
```

```python
#Step 3: Train a Linear Regression Model
#we use scikit learn to train the model
#the model will automatically learn: slop(m) and intersect(c)
model = LinearRegression()
model.fit(hours, marks)
```

```python
#Step 4: View the learned Parameters
#The regression line equation is: (y = mc + c), where, m = slope, c = intercept
print(f"Slope(m): {model.coef_[0]}")
print(f"Intercept(c): {model.intercept_}")

#So the equtaion is : marks = 4.3 X hours + 44.8
```

```python
# Step 5: Visualize the best fit line
plt. scatter(hours, marks, color = "blue", label = "Actual Data")

predicted_marks = model.predict(hours)
plt.plot(hours, predicted_marks, color = "red", label = "Best Fit Line")

plt.xlabel("Hourse")
plt.ylabel("Marks")
plt.title("Study Hours vs Marks for Linear Regression")
plt.legend()
plt.grid(True)
plt.show()
```

```python
#Predict Mark for a new Student
new_hours = np.array([[7]])
prediction = model.predict(new_hours)
print(f"Predicted Mark for {new_hours} hours: {prediction[0]}")
```
