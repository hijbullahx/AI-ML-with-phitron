## Section 6 — Distributions: Histograms and Boxplots for Numeric Features

Visualize numeric feature distributions and look for skewness or outliers.

```python
# 2. Plot boxplots for numeric features to detect outliers
import math

cols = [c for c in numeric_cols ]

n = len(cols)
rows = math.ceil(n/4)

plt.figure(figsize=(20, 4*rows))
for i, col in enumerate(cols, 1):
    plt.subplot(rows, 4, i)
    df.boxplot(column=col)
    plt.title(col, fontsize=10)
    plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

## Section 10 — Pairplot and Correlation Analysis

Explore relationships between features and correlations.


```python
# TODO:
# 1. Create a pairplot with selected numeric features plus the target
sns.pairplot(df[["area_se", "texture_se", "concavity_se", "radius_mean", "diagnosis"]], hue="diagnosis", diag_kind = "hist")
plt.show()
```

```python
# Convert diagnosis to numeric (M=1, B=0)
df['diagnosis_num'] = df['diagnosis'].map({'M':1, 'B':0})

# Generate correlation matrix
# Corrected: Removed the redundant 'diagnosis_num' from the list of columns passed to .corr()
corr_matrix = df[["area_se", "texture_se", "concavity_se", "radius_mean", "diagnosis_num"]].corr()

# Visualize with heatmap
plt.figure(figsize=(12,10))
sns.heatmap(corr_matrix, annot=True, cmap="coolwarm", fmt=".2f", linewidths=1)
plt.title("Correlation Heatmap for Numerical Features and Target")
plt.tight_layout()
plt.show()
```

```python
# Compute correlations of all numeric features with the target
corr_with_target = df[numeric_cols].corrwith(df['diagnosis_num'])

# Sort values
corr_with_target.sort_values(ascending=False)
```
