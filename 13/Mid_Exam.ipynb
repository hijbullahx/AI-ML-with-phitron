```python
# You must run this code and use that data set for all the questions
import numpy as np
import pandas as pd

np.random.seed(42)

ids = np.arange(1, 11)
ages = np.random.randint(18, 60, 10)
salaries = np.random.randint(30000, 90000, 10)
departments = np.array(["HR", "IT", "Finance", "IT", "HR", "Sales", "Finance", "IT", "Sales", "HR"])

DF = pd.DataFrame({
"id": ids,
"age": ages,
"salary": salaries,
"dept": departments
})

DF.to_csv("employees.csv", index=False)
print("Sample Data Created and Saved as employees.csv")
```

# 1) NumPy

## Ans to the Q No.1

```python
arr_ages = np.array(ages)
arr_salaries = np.array(salaries)

print(f"Array of Ages: \n{arr_ages}")
print(f"dtype: {arr_ages.dtype}")
print(f"ndim: {arr_ages.ndim}")
print(f"Shape: {arr_ages.shape}")
print(f"Size: {arr_ages.size}")

print(f"\nArray of Salaries: \n{arr_salaries}")
print(f"dtype: {arr_salaries.dtype}")
print(f"ndim: {arr_salaries.ndim}")
print(f"Shape: {arr_salaries.shape}")
print(f"Size: {arr_salaries.size}")
```

## Ans to the Q No.2

```python
highest_salary = np.max(arr_salaries)
lowest_salary = np.min(arr_salaries)
print(f"Heighest Salary : {highest_salary}")
print(f"Lowest Salary : {lowest_salary}")

avg_salary = np.mean(arr_salaries)
avg_age = np.mean(arr_ages)
print(f"Average Salary : {avg_salary}")
print(f"Average Age: {avg_age}")
```

## Ans to the Q No.3


```python
ags_grtr_30 = np.where(arr_ages > 30)
print(f"Ages greater than 30: {arr_ages[ags_grtr_30]}")
print(f"{len(arr_ages[ags_grtr_30])} employees are older than 30")
```

## Ans to the Q No.4

```python
increase_age_5 = arr_ages + 5
print(f"Ages after increasing by 5 years: {increase_age_5}")
```

## Ans to the Q No.5

```python
total_salary_expns = np.sum(arr_salaries)
print(f"Total Salary Expense: {total_salary_expns}")
diff_max_min_sal = np.max(arr_salaries) - np.min(arr_salaries)
print(f"Difference between highest and lowest salary: {diff_max_min_sal}")
```
